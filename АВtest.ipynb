**Задание 1**  
Имеется 2 выборки:
- первая выборка — это участки, обработанные агротехническим методом А;
- вторая выборка — участки, обработанные агротехническим методом Б.

Значения в выборках — это урожайность.
Требуется выяснить, является ли один из методов эффективнее другого, или различия урожайности обусловлены случайными факторами.
Для проверки гипотезы применить любой статистический критерий

`sample_a = [3352, 3080, 3195, 3448, 3373, 2804, 3190, 2969, 2979, 3082, 3028, 3290, 3152, 3024, 3088, 3066, 3298, 2959, 3062, 2829, 2489, 3130, 3172]`

и

`sample_b =
[3214, 3967, 3036, 3411, 3353, 3783, 3767, 3438, 3494, 3178, 2904, 3313, 3439, 3707, 3700, 3303, 3324, 3137, 3045, 2973, 3887, 3272, 3290, 3086, 3594, 2996, 3346, 3176, 3496, 3272]`




```python
# ваш код
# H0: Cредние значения урожайности двух методов одинаковы
# H1: Среднее значения урожайности двух методов различны

import numpy as np
from scipy.stats import mannwhitneyu
sample_a = [3352, 3080, 3195, 3448, 3373, 2804, 3190, 2969, 2979, 3082, 3028, 3290, 3152, 3024, 3088, 3066, 3298, 2959, 3062, 2829, 2489, 3130, 3172]

sample_b = [3214, 3967, 3036, 3411, 3353, 3783, 3767, 3438, 3494, 3178, 2904, 3313, 3439, 3707, 3700, 3303, 3324, 3137, 3045, 2973, 3887, 3272, 3290, 3086, 3594, 2996, 3346, 3176, 3496, 3272]

statistic, p_value = mannwhitneyu(sample_a, sample_b)

print("Статистика теста:", statistic)
print("p-value:", p_value)
alpha=0.05

if (p_value < alpha):
    print("Отвергаем нулевую гипотезу, различия урожайности статистически значимы ")
else:
    print("Не можем отвергнуть нулевую гипотезу")




```
