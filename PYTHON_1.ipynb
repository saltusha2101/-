**Задание 1**  
Пользователь вводит число. Напишите программу, которая осуществляет проверку:


1.   если число четное и отрицательное, то печатаем сообщение "отрицательное четное число"
2.   если число четное и положительное, то печатаем сообщение "положительное четное число"
3. если любое число нечетное , то печатаем сообщение "нечетное число"
4. если число = 0 , то печатаем сообщение "данное число равно 0"



````python
# ваш код
x = int(input('Введите число X:'))

if x < 0 and x % 2 == 0:
    print('Отрицательное четное число')
elif x > 0 and x % 2 == 0:
    print('Положительное четное число')
elif x % 2 != 0:
    print('Нечетное число')
else:
    print('Данное число равно 0')
````

**Задание 3**

Напишите программу-игру, в которой компьютер загадывает число от 1 до 100, а пользователь пытается его угадать. После  попытки пользователя программа должна сообщать, больше или меньше загаданное число, чем введенное пользователем.

Подсказка:
Для генерации случайного числа можно использовать модуль:
```
import random
random_number = random.randint(1,100)
```

````python
from random import randint

N = randint(1,100)
I = int(input("Угадайте целое число от 1 до 100:"))

while I != N:

  I = int(input("Повторите попытку:"))
  if I < N:
    print('Число меньше загаданного')
  elif I > N:
    print('Число больше загаданного')
  else:
    print('Вы угадали')
````
