**Задание 1**  
В группе из 20 студентов нужно выбрать 4 человека для участия в олимпиаде. Сколькими способами это можно сделать?

```python
# ваш код
def factorial(n):
    f = 1
    for x in range(1, n+1):
        f *= x
    return f
```

```python
def C_n_m(n, m):
    return int(factorial(n) / (factorial(n-m) * factorial(m)))
```

```python
C_n_m(20,4)
```

**Задание 2**  
Сколько различных трехзначных чисел можно составить из цифр 1, 2, 3, 4, 5, если цифры не повторяются?

```python
# ваш код

def A(n, m):
    return int(factorial(n) / factorial(n - m))
```

```python
A(5,3)
```

**Задание 3**  
Группу из 20 студентов нужно разделить на 3 бригады, причем в первую бригаду должны входить 3 человека, во вторую — 5 и в третью — 12. Сколькими способами это можно сделать?

```python
# ваш код

def groups(n, m):
    return int(factorial(n) / (factorial(n-m) * factorial(m)))

g1 = groups(20,3)
g2 = groups(17,5)
g3 = groups(12,12)
result = g1*g2*g3
print(result)


```
