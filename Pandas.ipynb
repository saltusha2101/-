**Задание 6**

- для числовых признаков рассчитайте долю пропусков
- заполните пропуски средним значением



```python
df.select_dtypes(include = ('float64','int64')).count().sort_values()  # рассчет доли пропусков числовых признаков
mean = df.select_dtypes(exclude='object').describe().loc['mean'] # определение среднего значения
mean
df_commbine = df.fillna(mean)  # заполние пропусков числовых признаков, средним значением
df_commbine
```

```python
df_commbine.isna().sum()
```

**Задание 7**

- для категориальных признаков рассчитайте долю пропусков
- заполните пропуски модой



```python
df.select_dtypes(include = ('object')).count().sort_values() #рассчет доли пропусков для категориальных признаков
most_frequent_objects = df.select_dtypes(include='object').describe().loc['top']
most_frequent_objects
df_commbine = df_commbine.fillna(most_frequent_objects)
df_commbine
```

```python
df_commbine.isna().sum()
```
