**Задание 1**  
Имеется выборка наблюдений некоторой случайной величины:

```[114.2, 114.0, 132.5, 141.8, 107.0, 126.9, 85.7, 149.6, 145.5, 130.7, 88.5, 111.8, 187.7, 102.2, 117.6, 113.5, 144.4, 104.4, 137.4, 149.8, 111.7, 121.9, 96.5, 138.7, 113.1, 126.0, 129.9, 138.5, 110.8, 129.4, 203.1, 124.1, 121.5, 126.9, 114.0, 128.0, 93.2, 116.6, 113.3, 122.9, 115.5, 78.7, 105.3, 113.2, 135.8, 121.1]```

Требуется вычислить среднее значение, медиану, выборочную дисперсию и межквартильный размах.



````python
# ваш код

import numpy as np
from scipy import stats
sample = np.array([114.2, 114.0, 132.5, 141.8, 107.0, 126.9, 85.7, 149.6, 145.5, 130.7, 88.5, 111.8,
                   187.7, 102.2, 117.6, 113.5, 144.4, 104.4, 137.4, 149.8, 111.7, 121.9, 96.5, 138.7,
                   113.1, 126.0, 129.9, 138.5, 110.8, 129.4, 203.1, 124.1, 121.5, 126.9, 114.0, 128.0, 93.2,
                   116.6, 113.3, 122.9, 115.5, 78.7, 105.3, 113.2, 135.8, 121.1])
print('Среднее значение:', np.mean(sample))
print('Медиана:', np.median(sample))
print("Выборочная дисперсия :", np.var(sample, ddof=0))
print("Межквартильный размах:", np.percentile(sample, 25) -np.percentile(sample, 75) )
````

**Задание 2**  
для выборки из задания 1 найти выбросы с помощью интерквантильного размаха

````python
# ваш код

Q1 = np.percentile(sample, 25)
Q3 = np.percentile(sample, 75)

IQR = Q3 - Q1
print("Интерквантильный размах:", IQR )

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = sample[(sample < lower_bound) | (sample > upper_bound)]
print("Выбросы:", outliers)
````
