```python
import pandas as pd
import numpy as np
```

**Задание 1**  

Создайте Pandas DataFrame с указанными параметрами:

**Условия:**
1. Количество строк: ( n = 50 ).
2. Колонки:
   - `client_id`: последовательность чисел от 1 до \( n \).
   - `age`: случайное равномерное распределение от 30 до 70.
   - `children_count`: случайное целое число от 0 до 4.
   - `purchases_count`: случайное целое число от 0 до 10.

**Требования:**
- Используйте `numpy` или `pandas` для генерации случайных данных.
- Запишите результат в переменную `df`.
- Выведите первые 5 строк с помощью `df.head()`.



```python
# ваш код

n = 50

dict  = {'client_id': np.arange (1, n + 1),
         'age': np.random.uniform (30,70,n),
         'children_count' : np.random.randint (0,4,n),
         'purchases_count' : np.random.randint (0,10,n)}

df  = pd.DataFrame(dict)
df.head(10)
```

**Задание 2**

1. Вычислите средний возраст клиентов для всего DataFrame: $ \frac{\sum \text{age}}{n} $.
2. Найдите количество клиентов без детей: $ \sum (\text{children_count} = 0) $.
3. Определите, сколько клиентов совершили более 5 покупок: $ \sum (\text{purchases_count} > 5) $.


```python
# ваш код
mean_age = df['age'].mean()
print("Средний возраст:",mean_age)
no_children = (df['children_count'] == 0).sum()
print('Клиенты без детей:',no_children)
purchases_count = (df['purchases_count'] > 5).sum()
print("Более 5 покупок совершили:",purchases_count)
```
