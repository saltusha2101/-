```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

**Задание 1**  

Загрузите данные `penguins` в DataFrame c помощью `sns.load_dataset`


```python
# ваш код
import seaborn as sns

df = sns.load_dataset("penguins")
df.head()
```

**Задание 2**

Постройте график, который будет разбит на 4 части и в каждой части будут гистограммы для следующих перемменных:

- bill_length_mm
- bill_depth_mm
- flipper_length_mm
- body_mass_g



```python
# ваш код
f, ax = plt.subplots(2, 2, figsize=(10, 10))
ax[0, 0].hist(df['bill_length_mm'], bins=20, color='lightgreen', edgecolor='black')
# ax[0, 0].plot([1, 2, 3, 4, 5, 6])
ax[0, 0].set_title('bill_length_mm')
ax[0, 0].set_xlabel('Распределение длины клюва')
ax[0, 0].set_ylabel('Частота')

ax[0, 1].hist(df['bill_depth_mm'], bins=20, color='lightgreen', edgecolor='black')
ax[0, 1].set_title('bill_depth_mm')
ax[0, 1].set_xlabel('Глубина клюва')
ax[0, 1].set_ylabel('Частота')

ax[1, 0].hist(df['flipper_length_mm'], bins=20, color='lightgreen', edgecolor='black')
ax[1, 0].set_title('flipper_length_mm')
ax[1, 0].set_xlabel('Длина плавника')
ax[1, 0].set_ylabel('Частота')

ax[1, 1].hist(df['body_mass_g'], bins=20, color='lightgreen', edgecolor='black')
ax[1, 1].set_title('body_mass_g')
ax[1, 1].set_xlabel('Масса тела')
ax[1, 1].set_ylabel('Частота')

# Регулировка размещения подграфиков для предотвращения перекрытия
plt.tight_layout(rect=[0, 0, 1, 0.96])

# Отображение графика
plt.show()
```

**Задание 3**

Постройте график body_mass_g в зависимости от species



```python
# ваш код
plt.figure(figsize=(10, 6))

sns.boxplot(x='species', y='body_mass_g', data=df, palette='Set3')
plt.title('Body Mass by Species')
plt.ylabel('Масса пингвина')
plt.xlabel('Вид пингвина')

plt.show()
```
